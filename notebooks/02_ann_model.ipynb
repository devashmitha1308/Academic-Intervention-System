```python

from pathlib import Path
import subprocess
import sys

PROJECT = Path.cwd() if (Path.cwd() / 'models').exists() else Path.cwd().parent
SCRIPT = PROJECT / "src" / "train_ann.py"

result = subprocess.run(
    [sys.executable, str(SCRIPT)],
    cwd=str(PROJECT),
    capture_output=True,
    text=True
)

print(result.stdout)

if result.stderr:
    print("Warnings or errors:")
    print(result.stderr)

print("Return code:", result.returncode)
```
